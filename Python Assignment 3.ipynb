```python
# 1 . Number Guessing Game

import random


secret_number = random.randint(1, 10)   # Generate random number between 1 and 10


attempts = 3    # Maximum number of valid guesses allowed

while attempts > 0:
    guess = int(input("Guess the number (between 1 and 10): "))


    if guess < 1 or guess > 10:
        print("Your guess is out of range. Please guess a number between 1 and 10.")
        continue

    attempts -= 1

    if guess == secret_number:
        print("Congratulations! You guessed the correct number.")
        break
    elif guess > secret_number:
        print("Too high. Try again.")
    else:
        print("Too low. Try again.")
else:

    print("Better luck next time!")      # Runs only if the loop ended without a break
    print("The number was", secret_number)
```

```python
number = int(input("Enter the number for which you want the multiplication table: "))

for i in range(1, 11):
    result = number * i
    print(f"{number} x {i} = {result}")   #The f before the opening quote tells this is an f-string.
                                           # Anything inside { } is treated as a Python expression, and its value inserted is converted into the text.
```

```python
def calculate_bmi(weight, height):

    return weight / (height ** 2)      #Return BMI = weight (kg) / height (m) squared.


weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

bmi = calculate_bmi(weight, height)
print(f"Your BMI is: {bmi:.2f}")   #.2f means 2 digits after the point, as a float.
```
